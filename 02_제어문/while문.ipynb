```python
# while 문의 기본구조
# while 조건문:
#     수행할 문장1
#     수행할 문장2
#     ...

treehit = 0

while treehit < 10:
    treehit = treehit + 1 # treehit += 1
    print(f"나무를 {treehit}번 찍었습니다")
    if treehit == 10:
        print("나무가 넘어갑니다.")
```

```python
coffee = 3
money = 300
while money:
    print("커피의 가격은 100원입니다.")
    coffee -= 1
    money -= 100
    print(f"남은 커피는 {coffee}개입니다.")
    print(f"남은 돈는 {money}입니다.")
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.")
        break
    elif money == 0:
        print("돈이 없습니다. 판매를 중지합니다.")
        break
```

```python
coffee = 2

while True:
    money = (int(input("돈을 넣어주세요: ")))

    if money == 300:
        print("커피를 줍니다.")
        coffee -= 1
        print(f"남은재고 {coffee}개 입니다.")

    elif money > 300:
        print(f"커피를 줍니다. 거스름돈 {money - 300}을 돌려줍니다.")
        coffee -= 1
        print(f"남은재고 {coffee}개 입니다.")
    else:
        print("돈을 다시 돌려주고, 커피를 주지 않습니다.")
        print(f"남은재고 {coffee}개 입니다.")
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.")
        break
```

```python
## while 문의 맨 처음으로 돌아가기 - continue

a = 0
while a < 10:
    a += 1
    if a % 2 == 0:
        continue
    print(a)
```
