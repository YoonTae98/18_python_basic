```python
# 3. % 연산자는 무엇을 구하는 연산자인지 설명하고, 실생활에서 활용할 수 있는 예시를 하나 들어 쓰시오.
## % 연산자는 나눗셈 후에 나머지를 반환

week = ["월", "화", "수", "목", "금", "토", "일"]

today = input("오늘은 무슨 요일인가요? (월/화/수/목/금/토/일): ")
days = int(input("몇 일 후의 요일을 계산할까요?: "))

today_num = week.index(today)
result = (today_num + days) % 7

print(days, "일 후는", week[result], "요일입니다.")

## 실생활에서 %7을 통한 요일계산을 할 수 있다
```

```python
'''
6. 원의 넓이를 계산하는 프로그램을 작성하시오.
- 반지름(r)은 변수로 저장 (예: r = 5)
- 원주율(pi)은 3.14159를 사용
- 공식: 넓이 = π × r²
'''
r = 5
pi = 3.14159
a = pi * (r**2)
print(a)
```

```python
'''
7. 복리 이자를 계산하는 프로그램을 작성하시오.
- 원금(principal): 1,000,000원
- 연이율(rate): 5% (0.05)
- 기간(years): 3년
- 공식: 최종금액 = 원금 × (1 + 연이율)^기간
'''
p = 1000000
r = 0.05
y = 3
a = p * (1 + r)**3
print(a)
```
