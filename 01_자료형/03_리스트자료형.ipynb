```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, ... ]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["life", "is"]]
e = [1, 2, "life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트의 인덱싱
a = [1, 2, 3]
print(type(a))
print(a[1])
print(a[2])
print(a[-1])

a = [1, 2, 3, ["a", "b", "c"]]
print(len(a))
print(a[-1][1]) #chaining
```

```python
## 리스트의 슬라이싱
a = [1, 2, 3, 4, 5]
a[1:4]
```

```python
a = [1, 2, 3, ["a", "b", "c"]]
a[3][1:3]
```

```python
## 리스트 연산
a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)
# print(a * b) 오류코드
print(a * 3)
```

```python
## 리스트 길이 구하기 - len
a = [1, 2, 3]
len(a)
```

```python
## 리스트의 수정과 삭제
a = [1, 2, 3]
a[2] = 4
print(a)
```

```python
a = [1, 2, 3]
del a[1]
print(a)
```

```python
## 리스트 관련 함수

## 리스트에 요소 추가하기 - append
a = [1, 2, 3]
a.append(4)
print(a)
a.append(["a", "b"])
print(a)
```

```python
## 리스트정렬 - sort
a = [1, 4, 3, 2]
a.sort()
print(a)

b = ["a", "c", "b"]
b.sort()
print(b)
```

```python
## 리스트 뒤집기 - reverse
a = ["a", "c", "b"]
a.reverse()
a
```

```python
## 인덱스 반환 - index
a = [1, 2, 3, 4]
a.index(3)
```

```python
## 리스트에 요소 삽입 - insert
a = [1, 2, 3]
a.insert(0, 4)
print(a)
```

```python
## 리스트에 요소제거 - remove()
a = [1, 2, 3, 1, 2, 3]
a.remove(2)
a
```

```python
## 리스트 요소 꺼내기 - pop
a = [1, 2, 3]
number = (a.pop()) # 비어있을땐 -1이 기본값

print(number)
print(a)
```

```python
## 리스트에 포함된 요소 x의 개수 세기 - count
a = [1, 2, 3, 4, 3]

a.count(3)
```

```python
## 리스트 확장
a = [1, 2, 3]
a.extend([4, 5])
a
```
