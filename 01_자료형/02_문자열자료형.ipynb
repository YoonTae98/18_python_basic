```python
# 문자열은 어떻게 만들고 사용할까?

## 1. 큰 따옴표
print("Hello World")

## 2. 작은 따옴표
print('python is fun')

## 3. 큰따옴표 3개를 연속으로 감싸기
print("""Life is too short, You need python""")

## 4. 작은따옴표 3개를 연속으로 감싸기
print('''Life is too short, You need python''')
```

```python
print("Chulsu's favorite food is kimchi")
print('Chulsu's favorite food is kimchi')
```

```python
## 문자열에 큰따옴표 표현하기
print('"python is very fun" chulsu says')
```

```python
## 역슬래시를 사용해서 작은따옴표, 큰따옴표 표시
print('Chulsu\'s favorite food is kimchi')
print("\"python is very fun\" chulsu says")
```

```python
multilines = """
Life is too short, 
You need Python
"""
print(multilines)
```

```python
## 문자열 연산하기
a = "Python "
b = "is fun"

a + b
# "Python" + "is fun"
```

```python
a * 3
```

```python
## 문자열 길이 구하기 len()

a = "Life is too short"
len(a)
```

```python
## 문자열 인덱싱과 슬라이싱
a = "Life is too short"
a[3]
a[16]
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a[5:7]
print(a[12:17])
a[-5:-1]
a[-5:]
a[:4]
```

```python
## 문자열 포매팅
print("I eat %d apples" %3)
```

```python
print("i eat %s apples" % "five")
```

```python
a = 3
b = "five"

print("I eat %d apples" % a)
print("i eat %s apples" % b)
```

```python
## format() 함수를 사용한 포매팅
"I eat {0} apples". format(3)
"I eat {0} apples". format("five")
```

```python
## 2개 이상의 값을 넣을때
num = 10
day = "three"
"I ate {0} apples, so i was sick for {1} days".format(num, day)
```

```python
## f문자열 포매팅
name = "홍길동"
age = 30

print(type(name))
print(type(age))

print(f"나의 이름은 {name}이고, 나이는 {age}입니다.")
```

```python
## 문자열 관련 함수들
a = "hobby"
type(a)

## 문자 개수 세기 - count

print(a.count("b"))
print("hobby".count("b"))
```

```python
## 위치 알려주기 - find()
a = "Python is the best choice"
print(a.find("b"))
print(a.find("k"))
```

```python
## 위치 알려주기 - index()
a = "Life is too short"
print(a.index("t"))
# print(a.index("k")) 에러코드
```

```python
## 문자열 삽입 - join
",".join(["a", "b", "c", "d"])
```

```python
## 소문자를 대문자로 바꾸기 - upper
a = "hi"
a.upper()
```

```python
a = "HI"
a.lower()
```

```python
## 양쪽 공백 지우기 - strip
a = "   hi   "
print(a)
print(a.strip())
```

```python
## 문자열 바꾸기 - replace
a = "Life is too short"
a.replace("Life", "Leg")
```

```python
## 문자열나누기 = split
a = "Life is too short"
a.split()
```
