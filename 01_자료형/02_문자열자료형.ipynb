```python
# 문자열은 어떻게 만들고 사용할까?

## 1. 큰 따옴표
print("Hello World")

## 2. 작은 따옴표
print('python is fun')

## 3. 큰따옴표 3개를 연속으로 감싸기
print("""Life is too short, You need python""")

## 4. 작은따옴표 3개를 연속으로 감싸기
print('''Life is too short, You need python''')
```

```python
print("Chulsu's favorite food is kimchi")
print('Chulsu's favorite food is kimchi')
```

```python
## 문자열에 큰따옴표 표현하기
print('"python is very fun" chulsu says')
```

```python
## 역슬래시를 사용해서 작은따옴표, 큰따옴표 표시
print('Chulsu\'s favorite food is kimchi')
print("\"python is very fun\" chulsu says")
```

```python
multilines = """
Life is too short, 
You need Python
"""
print(multilines)
```

```python
print('"python is very fun"
chulsu says')
```

```python
print('"python is very fun" \nchulsu says')
```

```python
## 문자열 연산하기
a = "Python "
b = "is fun"

a + b
# "Python" + "is fun"
```

```python
a * 3
```

```python
## 문자열 길이 구하기 len()

a = "Life is too short"
len(a)
```

```python
## 문자열 인덱싱과 슬라이싱
a = "Life is too short"
a[3]
a[16]
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a[5:7]
print(a[12:17])
a[-5:-1]
a[-5:]
a[:4]
```
